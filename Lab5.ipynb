### Create Maze

```python
from src.mazeData import makeMaze

n=10

maze1, init1, goal1, food1 = makeMaze(n, False) # No ghosts
maze1
```

### Create Transition Model

```python
from src.mazeData import defineMazeAvailableActions
from src.mazeData import makeMazeTransformationModel

mazeAvalActs1 = defineMazeAvailableActions(maze1)

mazeTM1 = makeMazeTransformationModel(mazeAvalActs1) 
```

### Create Maze Graph

```python
from src.maze2025GraphClass import mazeGraph
from src.mazeData import mazeStatesLocations

mazeWorldGraph1 = mazeGraph(mazeTM1, mazeStatesLocations(list(mazeTM1.keys())))
```

### Visualize Maze

```python
from src.mazeData import draw_maze

print(maze1)
draw_maze(maze1)
```

### Create Euclidian and Manhattan Agents

```python
from src.agents import ProblemSolvingPacmanAgentAstarEuclidian, ProblemSolvingPacmanAgentAstarManhattan
from src.task.MazeEnvironment import MazeEnvironment
```

Interactive Toggle

```python
interactive = False
```

### Run Simulation For Euclidian

```python
# Hanna's Euclidian distance agent
Euc_Agent = ProblemSolvingPacmanAgentAstarEuclidian(init1, mazeWorldGraph1, goal1, food1.copy())

mazeEnv = MazeEnvironment(mazeWorldGraph1, maze1.copy())

draw_maze(maze1, Euc_Agent.state)

mazeEnv.add_thing(Euc_Agent)

mazeEnv.run(999, interactive)
```

### Run Simulation For Manhattan

```python
Man_Agent = ProblemSolvingPacmanAgentAstarManhattan(init1, mazeWorldGraph1, goal1, food1.copy())

mazeEnv = MazeEnvironment(mazeWorldGraph1, maze1.copy())

draw_maze(maze1, Man_Agent.state)

mazeEnv.add_thing(Man_Agent)

mazeEnv.run(999, interactive)
```

```python
neighbors = {}

'''
Keys will be tuples representing (row, col)
From (1, 1) to (9, 9)
'''

from src.task2utils import sameCol, sameRow, sameHouse, asteriskNeighbours

for i in range(1, 10):
  for j in range(1, 10):
    key = (i, j)
    neighbors[key] = set(sameCol(key) + sameRow(key) + sameHouse(key) + asteriskNeighbours(key))

neighbors
```

```python
from src.task2utils import given, allVals

domains = {}

for i in range(1, 10):
  for j in range(1, 10):
    key = (i, j)
    domains[key] = allVals

# Set Pre-filled values
for key in given:
  domains[key] = given[key]

domains
```

```python
from src.CSPS import AsteriskSudokuCSP

aSudokuCSP = AsteriskSudokuCSP(domains, neighbors)

from src.algorithms import backtracking_search

result = backtracking_search(aSudokuCSP)
result
```

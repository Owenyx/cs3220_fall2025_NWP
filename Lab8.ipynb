# Lab 8 - Course Scheduling CSP with Local Search

### State Space Structure:
- 15 variables: 3 for each day, each representing a time slot
- The days of the variables are encoded as 1-5
- The time slots are encoded as a-c
- For example: The second time slot on Thursday would be 4b

- The values in the domains are like PPM-lc, PPM-lb, A1-lc, Empty

```python
variables = []

for day in range(1, 6):
  for time in ['a', 'b', 'c']:
    variables.append(f'{day}{time}')

variables
```

```python
course_items = [
  'Empty',
  'PPM-lc',
  'PPM-lb',
  'A1-lc',
  'A1-lb',
  'OS-lc',
  'OS-lb',
  'DB-lc',
  'DB-lb'
]
```

- Neighbors are time slots on same or adjacent days

```python
neighbors = {}

for varA in variables:
  for varB in variables:
    if (varA == varB):
      continue

    dayA = int(varA[0])
    dayB = int(varB[0])

    if abs(dayA - dayB) <= 1:
      neighbors.setdefault(varA, []).append(varB)

neighbors
```

```python
amountsDict = {
  'Empty': 2,
  'PPM-lc': 2,
  'PPM-lb': 2,
  'A1-lc': 2,
  'A1-lb': 1,
  'OS-lc': 2,
  'OS-lb': 1,
  'DB-lc': 2,
  'DB-lb': 1
}
```

```python
from src.CSPS import CourseSchedulingCSP

schedulingCSP = CourseSchedulingCSP(course_items, neighbors, amountsDict)
```

```python
from src.algorithms import min_conflicts1

solution = min_conflicts1(schedulingCSP)
solution
```

### Final Solution

```python
def print_schedule(schedule):
    slots = ['a', 'b', 'c']
    days  = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri']

    print(f"{'':6}{'Class1':8}{'Class2':8}{'Class3':8}")
    for day_index, day_name in enumerate(days, start=1):
        row = [schedule[f"{day_index}{slot}"] for slot in slots]
        print(f"{day_name:<6}" + "".join(f"{val:8}" for val in row))

print_schedule(solution)
```
